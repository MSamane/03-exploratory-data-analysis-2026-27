### 3.1.1 Finding potential correlations and data explanation
<a id="primer"></a>

Based on your code in Exercise 1, it would seem that the variables `mag` and `sig` are significantly correlated. This is not surprising, since [according to the USGS explanatory notes](https://earthquake.usgs.gov/data/comcat/data-eventterms.php#sig) `mag` is the magnitude of the event and `sig` is a number describing how significant the event is. Larger numbers indicate a more significant event. This value is determined on several factors, including: magnitude, maximum MMI, felt reports, and estimated impact. Thus, one would expect magnitude and significance to be strongly correlated.

Since we are exploring this dataset, let's define the remaining terms (all according to the USGS):

`Gap`: largest azimuthal gap between azimuthally adjacent stations (in degrees)

`dmin`: Horizontal distance from the epicenter to the nearest station (in degrees). 1 degree is approximately 111.2 kilometers. In general, the smaller this number, the more reliable is the calculated depth of the earthquake.

`nst`: The total number of seismic stations used to determine earthquake location.

`rms`: The root-mean-square (RMS) travel time residual, in sec, using all weights. This parameter provides a measure of the fit of the observed arrival times to the predicted arrival times for this location. Smaller numbers reflect a better fit of the data. The value is dependent on the accuracy of the velocity model used to compute the earthquake location, the quality weights assigned to the arrival time data, and the procedure used to locate the earthquake.

`tsunami`: This flag is set to "1" for large events in oceanic regions and "0" otherwise. The existence or value of this flag does not indicate if a tsunami actually did or will exist. If the flag value is "1", the event will include a link to the NOAA Tsunami website for tsunami information. The USGS is not responsible for issuing Tsunami warnings; they are simply providing a link to the authoritative NOAA source.

`latitude` and `longitude`: coordinates

`depth`: The depth at which the earthquake begins to rupture. This is not very well constrained because of the different referencials for depth used by different stations.

`place`: the name of the location where the Earthquake happened: a string.

Ok, now that we know what each of these columns means, let's plot some data!

### 3.2.1 Figures and Subplots 
<a id="subplots"></a>
The power of `matplotlib` is its ability to control pretty much every part of a plot. You can find the [full API description of matplotlib and documentation here](https://matplotlib.org/), as the topic is too large to be fully covered in this session. That said, using `matplotlib` can be quite intractable, and there are a fair few people who might find it really frustrating! We'll go through it slowly in this section.

Indeed, we'll start with something that some might find incredibly abstract: the difference between a `figure` and an `axes`. You can think of `figure` as a canvas upon which multiple plots (`axes`) are drawn. 

In the cell below, we start by using `plt.subplots()` function. This function returns a `figure` object (which we assign to `fig`) and an array of `axes` objects (which we assign to `axes`). The `2, 2` in the parentheses tells the function that we want our plots to be arranged in 2 columns and 2 rows. Consequently, the returned `axes` objects will also be arranged a 2x2 array, which can be accessed the same way you would access a list of objects.

We can assign plots to each subplot (each `axes` object) by refering to them by name. For instance, we can plot the histogram of the distribution of earthquakes in red in the `axes` object contained in the [0,0] position (upper left), the significance in [0,1] in blue (upper right), the depth in [1,0] in green (lower left), and the longitude in [1,1] in black (lower right):


````python
fig, axes = plt.subplots(2, 2)
axes[0,0].hist(caliquakes_df['mag'], bins=70, color='r', alpha=0.5)
axes[0,1].hist(caliquakes_df['sig'], bins=70, color='b', alpha=0.5)
axes[1,0].hist(caliquakes_df['depth'], bins=70, color='g', alpha=0.5)
axes[1,1].hist(caliquakes_df['longitude'], bins=70, color='k', alpha=0.5)

plt.plot();
````

#### Adjusting the spacing around subplots

That looks good, but it's certainly not publication quality - there's a lot of missing information in the plots. Notably, the x- and y-axis labels are missing, there are no plot titles for each subplot (so readers will have no idea what data is in each subplot), and perhaps most obviously, the y-axis tick labels are bleeding into the other subplots! 

One way of solving the last issue is to use the same axis scale for all plots (assuming that this works for your collection of subplots). This enables us to tell matplotlib to only print the axis values once to avoid clutter. We can also adjust the spacing between subplots, reducing wasted space. If we set it to 0, the subplots touch:

````python
# note the use of the sharex and sharey parameters in the subplots call - this forces all plots in the figure to use the same 
# scale. Note that this only works in this case because we're plotting the same data on each subplot. 
fig, axes = plt.subplots(2, 2, sharex=True, sharey=True)
axes[0,0].hist(caliquakes_df['mag'], bins=70, color='r', alpha=0.5)
axes[0,1].hist(caliquakes_df['mag'], bins=70, color='b', alpha=0.5)
axes[1,0].hist(caliquakes_df['mag'], bins=70, color='g', alpha=0.5)
axes[1,1].hist(caliquakes_df['mag'], bins=70, color='k', alpha=0.5)
plt.subplots_adjust(wspace=0, hspace=0)
````

### 3.2.3 Ticks, Labels, and Legends
<a id="ticks"></a>

#### Setting the title, axis labels, ticks, and ticklabels

Let's make our figure a little more informative by changing the ticks, the label, the axis limits and the title.

````python
# Let's recreate the magnitude vs significance figure as a subplot of 1 figure

# Note that the following two commands are the same as calling fig, ax = plt.subplots(1,1) - it is merely a different way of 
# reaching the same endpoint.
fig = plt.figure()
ax = fig.add_subplot(1, 1, 1)

# Plot a scatterplot on the ax object.
# What do you think the label parameter does?
ax.scatter(caliquakes_df['mag'],caliquakes_df['sig'], label='Magnitude')

# Now let's tell matplotlib that we only want to see the following ticks:
ticks = ax.set_xticks([2, 4, 6, 8])

# Let's set a title for our figure:
ax.set_title('Magnitude vs significance')

# And appropriate labels for the x and y axes:
ax.set_xlabel('Earthquake Magnitude')
ax.set_ylabel('Significance')

# We can also place a legend. The position determines where the legend will be displayed. 'Best' lets 
# the library decide where to put it. 
ax.legend(loc = 'best')
plt.plot;
````

**Bonus Task:** Look at the documentation for ax.legend and try experimenting with different places you can put the legend. Is it possible to put the legend outside of the plot?

### Setting Axis Limits 

Let's focus on the data between magnitude 3 and 4 by setting an axis limit on the x axis:

````python
# Since we will use the plot above a few times, let's create a function that creates the plot for us:
# The function below also shows you how to set a title as well as x- and y-axis labels. 
def create_plot():
    fig = plt.figure()
    ax = fig.add_subplot(1, 1, 1)
    ax.scatter(caliquakes_df['mag'],caliquakes_df['sig'], label='Magnitude')
    ticks = ax.set_xticks([2, 4, 6, 8])
    ax.set_title('Magnitude vs significance')
    ax.set_xlabel('Earthquake Magnitude')
    ax.set_ylabel('Significance')
    ax.legend()
    return (fig, ax)

# Now we set the x and y axis limits to focus on the lower end of the scale:

fig, ax = create_plot()
ax.set_xlim(6,8)
ax.set_ylim(0,3000)
plt.plot;
````

Play with the x- and y-axis limits in the graph above. 

#### Annotations and Drawing on a Subplot

You can annotate on any subplot by using the following syntax:

```python
ax.text(x, y, 'Hello world!', 
        family='monospace', fontsize=10)
```

Note that the x and y values in the `ax.text()` function use the axis values as coordinates (i.e., if you want to annotate something in the middle of a plot that has x-axis values from (0,100) and y-axis values from (0,1000), you'd have to use x = 50 and y = 500). 

````python
# We can first reuse our plot creation function above. Notice that it returns a tuple with figure and axes. 
# So we can do this to assign a new figure and axes:
fig, ax = create_plot()

# Now let's set some reasonable axis limits for what I want to demonstrate:
ax.set_xlim(6,8)
ax.set_ylim(0,3500)

# And let's label the two outliers data point that have a significance >2500. Let's extract them from the dataframe:

outliers = caliquakes_df[caliquakes_df['sig']>2500] # Remember how to do that?

# Now let's add an annotation to the plot. We will use the name of the place as our label ('place' in the dataset).
# Note the use of the iterrows() function in pandas that returns both the index and the row of the dataframe
# We can use the values of the rows to set an x,y location and plot the data:

for index, row in outliers.iterrows():
    ax.text(row['mag']+.02,row['sig']+50,row['place'])
````

Now we know where these anomalously significant earthquakes happened!

### 3.3.3 Scatter Plot with Regression
<a id="snsscatter"></a>
We can, of course, plot scatter plots with Seaborn. In fact, the Seaborn library goes one step further: by default, Seaborn calculates and plots the regression line associated with the data given to the scatter plot. Here is the same scatter plot as before, i.e. magnitude vs significance of earthquakes, but plotted in seaborn:

````python
# Let's reset the seaborn parameters to values I actually like
sns.set_theme(style='whitegrid')

plt.figure(figsize=(10,6))
sns.regplot(x='mag', y='sig', data=caliquakes_df) # Notice how in Seaborn the data is an argument passed to the library
plt.title('Magnitude vs Significance of Earthquakes');
````

**Bonus Task**: It's actually pretty difficult to see the regression line in this plot. Look through the documentation and see if you can find out how to change the colour of the regression line while keeping the colour of the data points the same.

You'll notice that the regression line is skewed significantly toward lower `sig` values. Indeed, beyond magnitude 5, the regression line is a terrible predictor of `sig`. This is because most of the data in this dataset has a magnitude of <5, significantly skewing the orientation of the regression line. 

To fix this, Let's try to create a regression for the earthquakes with magnitude > 5.

````python
data = caliquakes_df[caliquakes_df['mag']>5]
plt.figure(figsize=(10,6))
sns.regplot(x='mag', y='sig',data=data)
plt.title('Magnitude vs Significance of Earthquakes');
````

Now it is much clearer that the regression was skewed by lower values (still is). Notice that we have even a confidence interval plotted on the regression. If you don't want the regression line, simply pass this argument to the function: 
```python 
fit_reg=False
``` 
Alternatively, you can also turn off the scatter data and only plot the regression:
```python
scatter=False
```
Another useful way to represent the data is to bin the x-axis values. This does not change the regression line, but it declutters the plot. You can also pass a `x_ci` parameter to display the error on each bin. Here, **ci** stands for 'confidence interval', but if you pass the "sd" as the value to this argument, it will show you the standard deviation of the values in that bin, allowing you to assess the variability of the data. Let's try it:

````python
plt.figure(figsize=(10,6))
sns.regplot(x='mag', y='sig', x_bins= 1500, x_ci="sd", data=caliquakes_df)
plt.title('Magnitude vs Significance of Earthquakes');
````

The example above shows 
1. Most of the earthquake data is concentrated below magnitude 4. Even when creating 1500 bins, all earthquakes with magnitudes greater than 6 fall into the final bin and are not plotted separately! I would need to create >10k bins to even see these larger earthquakes
2. The significance becomes increasingly harder to predict with increasing magnitude (the standard deviation of each bin is increasing).

## 3.4 Plotting Geo-referenced data
<a id="georef"></a>
You'll find that we often need to plot things over a map or an image in the Geosciences. The study of earthquakes is no exception! Let's see how we can do that.

First, let's define a DataFrame that contains all earthquake with magnitude > 6.7. We will plot these earthquakes on a map of California.

````python
bigquakes_df = caliquakes_df[caliquakes_df['mag']>6.7]
````

matplotlib has a module called 'image' that allows you to open images. I have also saved a png image of the state of california in the 'img' folder of this course. All we need to do is the following:

* Import the image module form matplotlib
* Import the specific image from the folder using image read
* Plot the image, specifying the axis dimensions of this image. In this case, this means the coordinates of the image edges, which are 124.55 to 113.80 degrees West, 32.45 to 42.05 degrees North
* Plot whatever else we want on top of this image.

Let's do it!

````python
# Read the image
california_img = plt.imread('img/california.png')

# Create a plot, in this case, a scatter plot of bigquakes_df. Plot the latitute and longitude as x-y coordinates, and use the magnitude times 10 as the size of the symbol
ax = bigquakes_df.plot(kind="scatter", x="longitude", y="latitude", figsize=(10,7),
                       s="sig", label="Magnitude", # use the significance for the size of the point and its colour
                       c="sig", cmap=plt.get_cmap("jet"), 
                       colorbar=False, alpha=0.4)

# Use imshow to plot the image
plt.imshow(california_img, extent=[-124.55, -113.80, 32.45, 42.05], alpha=0.5,
           cmap=plt.get_cmap("jet"))
plt.ylabel("Latitude", fontsize=14)
plt.xlabel("Longitude", fontsize=14)

# Create a color bar on the side, based on the significance value. This is optional, but nice!
mag = bigquakes_df["sig"]

# We can create tick values based on the min() and max() magnitude - we don't need to calculate these ourselves!
# Look up what np.linspace is doing or simply run this in a new cell to see what is going on
tick_values = np.linspace(mag.min(), mag.max(), 11)

# Create a colorbar
cbar = plt.colorbar()
cbar.ax.set_yticklabels(tick_values, fontsize=14)
cbar.set_label('Significance', fontsize=16)

plt.legend(fontsize=16);
````

### Saving figures
This is such a nice plot that we want to save it! This is really easy: simply use the 'savefig' method and pass a string argument of the path you want, a format, and resolution. Here, I will save it into my 'img' folder as 'california_big_quakes'. I chose the 'png' format but you can also choose 'jpeg' and many other formats. 

This is particularly useful for your MSci thesis or other reports where you need stunning graphics:

````python
plt.savefig('img/california_big_quakes', format='png', dpi=300);
````
